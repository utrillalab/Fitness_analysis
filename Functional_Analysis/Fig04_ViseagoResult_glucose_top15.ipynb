```python
def sum_pl3(my_list, condition):  #select a list with the file name and condition
    eco_db4691_final = "~/Notebooks/LIZ/COG/allgen4691_fitfg_newname.csv"
    proteomic = "~/Notebooks/LIZ/COG/proteo2346.csv"
    
    all_db =  pd.read_csv(eco_db4691_final)
    total_proteo = all_db.sum(numeric_only=True)
    df = pd.read_csv(my_list)
    df2 = pd.read_csv(proteomic)
    lista2 = list(df["genes"])
    matching_genes = all_db[all_db["gene"].isin(lista2)]
    PL = matching_genes[condition].sum(numeric_only=True)
    tgenes = len(lista2)
    mw_prote = df2[df2["gen"].isin(lista2)]
    genes_withpl = len(mw_prote)
    proteome_fraction = (PL / total_proteo[condition]) * 100
    
    return proteome_fraction
```

```python
df_func_ygen = pd.merge(df_PL, df_gen, on=['Categ', 'GO'], how='left')
PL_sort2 = df_func_ygen.sort_values(by=['Categ','PF'], ascending=False)
PL_sort2.head()
```

```python
PL_sort2 = PL_sort2[PL_sort2['GO'] != 'GO:0019058']
PL_sort2.head()
```

```python
result_df = pd.DataFrame(columns=['Categ', 'gene', 'Frequency'])

# Calcular la frecuencia de cada gen por categoría
for category in PL_sort2['Categ'].unique():
    # Filtrar por categoría
    category_df = PL_sort2[PL_sort2['Categ'] == category]
    
    # Obtener el conjunto de genes y contar frecuencias
    genes_list = category_df['genes'].str.split(';').explode()
    genes_counts = genes_list.value_counts().reset_index()
    genes_counts.columns = ['gene', 'Frequency']
    genes_counts['Categ'] = category
    
    result_df = pd.concat([result_df, genes_counts], ignore_index=True)
result_df
```

```python
proteomic_df = pd.read_csv('./infiles/proteomicf_2346.csv')
result_df = result_df.merge(proteomic_df[['gene', 'Glucose']], how='left', left_on='gene', right_on='gene')
result_df['Glucose'] = result_df['Glucose'].fillna(0)
result_df
```
