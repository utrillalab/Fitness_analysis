### Loading necessary dependencies

```python
import cobra
import pickle
import numpy as np
import pandas as pd
import json
import cobrame
import ecolime
```

### Labels, reactions and loading each carbon source ME solution (fluxes)

```python
# Previously generated ME solutions are labeled with their active uptake reaction
EX_reactions = ['EX_glyc_e', 'EX_gal_e', 'EX_man_e', 'EX_gam_e', 'EX_pyr_e', 'EX_fru_e',
                'EX_glc__D_e', 'EX_xyl__D_e', 'EX_succ_e', 'EX_ac_e']

# The carbon sources of each uptake reaction
carbon_sources = ['Glycerol', 'Galactose', 'Mannose', 'Glucosamine', 'Pyruvate', 'Fructose',
                  'Glucose', 'Xylose', 'Succinate', 'Acetate']

# Dictionary to contain the fluxes for each carbon source
fluxes_dict = {}

for c, c_reaction in enumerate(EX_reactions):
    with open('ME_Output/' + c_reaction + '_34_1_036.pkl', 'rb') as f:
        fluxes_dict[carbon_sources[c]] = pickle.load(f) # Loading each ME solution
    
    # Showing the loaded solution for each carbon source and the number or reactions
    print(carbon_sources[c], '\t', len(fluxes_dict[carbon_sources[c]].keys()), 'reactions')

# Showing all carbon sources
fluxes_dict.keys()
```

### Loading original model for the index to select necessary reactions

```python
with open('iJL1678b_ML_KO_keff_FbFP.pickle', 'rb') as f:
        me = pickle.load(f)
```

### Creating a data frame that contains the translation reactions of all the genes in the model in each selected carbon source

```python
# Creating data frame
df = pd.DataFrame()

# Creating a column with the genes present in the model
df['genes'] = [reaction.id.split('_')[1] for
               reaction in me.reactions if type(reaction) == cobrame.core.reaction.TranslationReaction]

# Selecting the translation flux for each gene in each carbon source
for c, carbon in enumerate(carbon_sources):
    df[carbon] = [fluxes_dict[carbon][reaction.id] for
                  reaction in me.reactions if type(reaction) == cobrame.core.reaction.TranslationReaction]

df = df.drop([1568, 1569], axis = 0) # Removing "dummy" reaction and an untranslated exogenous gene (both fluxes are 0)

# Showing data frame
df
```

### Exporting data frame with the fluxes

```python
Updated_ME_fluxes = df
Updated_ME_fluxes.to_csv('ME_Output/Updated_ME_fluxes.csv', index = None)
Updated_ME_fluxes.to_csv('../Input_Data/ME_Output/Updated_ME_fluxes.csv', index = None)
```
